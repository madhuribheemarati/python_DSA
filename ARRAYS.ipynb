GIVEN AN ARRAY AND AN INTEGER K,ROTATE THE ARRAY TO THE RIGHT BY K STEPS
arr=[1,2,3,4,5,6,7]
k=3


```python
def rotatearray(arr,k):
  k=k%len(arr)
  return arr[-k:] + arr[:-k]
arr=[1,2,3,4,5,6,7]
k=3
rotated_array=rotatearray(arr,k)
print(rotated_array)
```

REVERSE THE GIVEN ARRAY IN-PLACE ,MEANS WITHOUT USING ANY EXTRA DATA STRUCTURE
arr=[2,4,5,7,9,12]

```python
def reversearray(arr):
  start,end=0,len(arr)-1
  while start<end:
    arr[start],arr[end]=arr[end],arr[start]
    start+=1
    end-=1
arr=[2,4,5,7,9,12]
reversearray(arr)
print(arr)
```

GIVEN A SORTED ARRAY,REMOVE THE DUPLICATE ELEMENTS WITHOUT USING ANY EXTRA DATA STRUCTURE

arr=[1,1,2,2,2,3,3,4,4,4,5,5]

```python
def remove_duplicates(arr):
    unique_index = 0
    for i in range(1, len(arr)):
        if arr[i] != arr[unique_index]:
            unique_index += 1
            arr[unique_index] = arr[i]
    return arr[:unique_index + 1]

arr = [1, 1, 2, 2, 2, 3, 3, 4, 4, 4, 5, 5]
new_arr = remove_duplicates(arr)
print( new_arr)

```
